```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('tmdb_5000_movies.csv')
```

```python
gen = df['genres'].str.extractall(r'"name":\s*"([^"]*)"')[0]
```

```python
gen_count = gen.value_counts()
```

```python
sort = gen_count.sort_index()
```

```python
plt.figure(figsize=(24,12))
plt.bar(sort.index, sort.values)
plt.xlabel('Genre')
plt.ylabel('Number of Movies')
plt.title('Number of Movies per Genre')
plt.show()
```
