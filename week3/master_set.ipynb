```python
import pandas as pd
```

```python
movies = pd.read_csv('movies_metadata.csv', low_memory=False)
keywords = pd.read_csv('keywords.csv')
credits = pd.read_csv('credits.csv')
links = pd.read_csv('links.csv')
```

```python
movies['id'] = pd.to_numeric(movies['id'], errors='coerce')
movies = movies.dropna(subset=['id'])
movies['id'] = movies['id'].astype(int)
```

```python
links['tmdbId'] = pd.to_numeric(links['tmdbId'], errors='coerce')
links = links.dropna(subset=['tmdbId'])
links['tmdbId'] = links['tmdbId'].astype(int)

```

```python
keywords['id'] = pd.to_numeric(keywords['id'], errors='coerce')
credits['id'] = pd.to_numeric(credits['id'], errors='coerce')

keywords = keywords.dropna(subset=['id'])
credits = credits.dropna(subset=['id'])

keywords['id'] = keywords['id'].astype(int)
credits['id'] = credits['id'].astype(int)
```

```python
movies = movies.drop_duplicates(subset='id')
keywords = keywords.drop_duplicates(subset='id')
credits = credits.drop_duplicates(subset='id')
links = links.drop_duplicates(subset='tmdbId')
```

```python
master = movies.merge(links, left_on='id', right_on='tmdbId', how='left')
```

```python
master = master.merge(keywords, on='id', how='left')
```

```python
master = master.merge(credits, on='id', how='left')
```

```python
master = master.drop(columns=['tmdbId', 'movieId', 'imdbId'])
```

```python
print(master.columns)
print(master.shape)
```

```python
master.to_csv('master_dataset.csv', index=False)
```
